```python
# Create schema sai_schema
spark.sql("CREATE SCHEMA IF NOT EXISTS sai_schema")

# Create Employee delta table in sai_schema
spark.sql("""
CREATE TABLE IF NOT EXISTS sai_schema.Employee (
    employee_id INT,
    employee_name STRING,
    dept_name STRING
)
USING DELTA
""")

# Insert 10 rows of data
spark.sql("""
INSERT INTO sai_schema.Employee VALUES
    (1, 'Alice', 'HR'),
    (2, 'Bob', 'Finance'),
    (3, 'Charlie', 'IT'),
    (4, 'David', 'Marketing'),
    (5, 'Eve', 'Sales'),
    (6, 'Frank', 'Operations'),
    (7, 'Grace', 'Legal'),
    (8, 'Heidi', 'Engineering'),
    (9, 'Ivan', 'Support'),
    (10, 'Judy', 'Admin')
""")

display(spark.table("sai_schema.Employee"))
```
