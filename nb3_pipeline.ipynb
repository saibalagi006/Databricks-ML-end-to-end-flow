```python
dbutils.widgets.dropdown("env", "QA", ["QA", "prod"], "Environment")
dbutils.widgets.text("param_text", "", "Parameter Text")
```

```python
env = dbutils.widgets.get("env")
param_text = dbutils.widgets.get("param_text")
print(f"env is {env}")
print(f"param_text is {param_text}")
```

```python
dbutils.widgets.text("num1", "0", "Number 1")
dbutils.widgets.text("num2", "0", "Number 2")

def add_two_numbers():
    num1 = float(dbutils.widgets.get("num1"))
    num2 = float(dbutils.widgets.get("num2"))
    return num1 + num2

result = add_two_numbers()
print(f"Sum: {result}")
```

```python

dbutils.jobs.taskValues.set(
    key="addition_result",
    value=result
)
```
